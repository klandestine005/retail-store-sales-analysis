```python
#Removing Item Column
#df = df.drop('Item', axis=1)
```

```python
#Handling Quantity Column
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
df['Quantity'].isna().sum()
```

```python
#Handling Total Spent Column
df['Total Spent'] = df['Total Spent'].fillna(math.floor(df['Total Spent'].mean()))
df['Total Spent'].isna().sum()
```

```python
#Handling Price Per Unit Column
df['Price Per Unit'] = df['Total Spent'] / df['Quantity']
df['Price Per Unit'].isna().sum()
```

```python
#Handling Discount Applied Column
df['Discount Applied'] = df['Discount Applied'].fillna('Unknown')
df['Discount Applied'].isna().sum()
```
